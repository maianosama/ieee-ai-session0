```python
# ==========================================================
# Task 1: Create Your Student Profile Dictionary
# ==========================================================
student_profile = {
    "name": "Maian Osama",
    "faculty_major": "Computer Engineering - AASTMT Heliopolis",
    "ai_interest": "Deep Learning and Computer Vision"
}

# Print to verify
print("Student Profile Created:", student_profile)
```

```python
# ==========================================================
# Task 2: Build the Text Analyzer Function
# ==========================================================
def analyze_bio(bio_text):
    # Calculate total characters including spaces
    char_count = len(bio_text)

    # Split the text into words and count them
    word_count = len(bio_text.split())

    # Return both counts
    return char_count, word_count
```

```python
# ==========================================================
# Task 3: Test and Print Your Results
# ==========================================================

# 1. Write a 1-2 sentence bio describing your goals in AI
my_bio = "As a computer engineering student, I am eager to master artificial intelligence and computer vision to build smart systems that solve real-world challenges."

# 2. Call your `analyze_bio()` function on that string and unpack the results
total_chars, total_words = analyze_bio(my_bio)

# 3. Print a clean formatted summary using f-strings
print("=" * 40)
print("             STUDENT AI PROFILE")
print("=" * 40)
print(f"Name:          {student_profile['name']}")
print(f"Major:         {student_profile['faculty_major']}")
print(f"AI Interest:   {student_profile['ai_interest']}")
print("-" * 40)
print(f"Bio:           {my_bio}")
print("-" * 40)
print(f"Character Count: {total_chars}")
print(f"Word Count:      {total_words}")
print("=" * 40)
```
